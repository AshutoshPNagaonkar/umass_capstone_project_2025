    <b>I. Strategic Data Preparation (The Foundation) </b><br>
The project began with rigorous data governance and modeling.<br>
Data Loading & Validation (Process): Six dimension tables and a core fact table were loaded, demonstrating robust data handling with Pandas.<br>
Targeted Filtering (Process): The main dataset was strategically filtered to isolate the most relevant time frame (2015–2019) into a dedicated, clean DataFrame, ensuring high data quality for subsequent analytical tasks.<br>
<b>II. The Gen AI Interpretation Module</b><br>
This module is the core of the NLQ system, handling the complex translation from unstructured text to structured command.<br>
Technology Stack: The module is built using <i>Hugging Face Transformers</i> for high-performance natural language processing.<br>
Core Function: Function (generate_visual_json) is responsible for analyzing the user's prompt via a two-step process:<br>
Intent Classification (Process): It employs a Zero-Shot Classification model to categorize the user’s request (e.g., year wise entries in Texas) without requiring task specific training data.<br>
Dynamic Parameter Extraction (Process): Sophisticated keyword and regex logic extracts critical entities (e.g., specific years, year range, port, states, or measures) from the text.<br>
Output: The function's result is a standardized JSON command that precisely defines all required data preparation steps, chart configuration, filtering rules and sorting parameters.<br>
<b>III. The Visualization Execution Engine</b><br>
This module ensures the reliable and rapid display of the analytical result.<br>
Core Function: create_visual_from_json: This single function serves as the analytical engine, processing the structured JSON command generated by the AI module.<br>
Execution Processes: 
Filter Application: It iterates through the filter criteria in the JSON, dynamically applying necessary lookups (e.g., finding the numerical ID for a State name) before filtering the main dataset.<br>
Data Aggregation: It performs the specified grouping and aggregation (.groupby().sum()) to compute the final metric values.<br>
Visualization: It uses industry-standard libraries (Matplotlib and Seaborn) to render a high-quality bar chart, using the title and axis labels extracted from the JSON configuration. </span><br><br>
